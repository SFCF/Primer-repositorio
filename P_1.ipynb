# implementación de OLS

$$\theta_0=\bar{y}-\theta_1\bar{x}$$

$$\theta_1=\frac{\sum_{i=1}^{n}x^i(y^i-\bar{y})}{\sum_{i=1}^{n}x^i(x^i-\bar{x})}


```python
def T1 (x, y):
    num = 0
    den = 0
    for i in range (len (x)):
        num += x[i] * (y[i] - sts.mean(y))
        den += x[i] * (x[i] - sts.mean(x))
    return num / den
```

```python
def T0 (x, y, t1h):
    return sts.mean(y) - t1h * sts.mean(x)
```

```python
#OLS

t1h = T1(x, y_2)
t0h = T0(x, y_2, t1h)

print(t1h, t0h)
```

```python
plt.plot(x, y_2, '.k', label="data")
yh = t0h+t1h*x

plt.plot(x, yh, '-r', label="model")
plt.legend()
plt.show()
```
